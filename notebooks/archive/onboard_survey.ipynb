# ארכיון — ניתוח סקר נוסעים היסטורי (`trips.csv`)

ניתוח חד-פעמי של סקר עליה-לאוטובוס (onboard) שנשמר ב-`data/raw/trips.csv`. נשמר לארכיון לצורך שחזור; אינו חלק מהפייפליין החי.

```python
import pandas as pd
trips = pd.read_csv('../data/raw/trips.csv')
print(trips.shape)
trips.head()
```

## עומס שבועי לפי קו

```python
cols = [c for c in ['line_id','sign','weekly_passengers','daily_passengers',
                    'avg_commuters_per_ride_weekly'] if c in trips.columns]
trips[cols].head(20)
```

## קווים בעומס נמוך (`low_demand_flag`)

```python
if 'low_demand_flag' in trips.columns:
    display(trips['low_demand_flag'].value_counts())
    display(trips.loc[trips['low_demand_flag']==1, cols].head(20))
else:
    print('no low_demand_flag column')
```
